## TP 1 — Algorithmes de tri

```python
def tri_selection(lst):
    """Tri par selection — O(n²)."""
    lst = lst.copy()
    n = len(lst)
    for i in range(n - 1):
        idx_min = i
        for j in range(i + 1, n):
            if lst[j] < lst[idx_min]:
                idx_min = j
        lst[i], lst[idx_min] = lst[idx_min], lst[i]   # echange
    return lst

notes = [12, 15, 8, 17, 11, 14, 6]
print(tri_selection(notes))
print(notes)   # original inchange ?
```

```python
def tri_insertion(lst):
    """Tri par insertion — O(n²)."""
    lst = lst.copy()
    for i in range(1, len(lst)):
        cle = lst[i]
        j = i - 1
        while j >= 0 and lst[j] > cle:
            lst[j + 1] = lst[j]
            j -= 1
        lst[j + 1] = cle
    return lst

print(tri_insertion(notes))
print(sorted(notes))   # Timsort Python
```

## TP 2 — Recherche

```python
def recherche_sequentielle(lst, valeur):
    """Recherche lineaire — O(n)."""
    for i, v in enumerate(lst):
        if v == valeur: return i
    return -1

def recherche_dichotomique(lst, valeur):
    """Recherche dichotomique — O(log n). Liste triee obligatoire."""
    gauche, droite = 0, len(lst) - 1
    while gauche <= droite:
        milieu = (gauche + droite) // 2
        if lst[milieu] == valeur: return milieu
        elif lst[milieu] < valeur: gauche = milieu + 1
        else: droite = milieu - 1
    return -1

notes_triees = sorted(notes)
print(recherche_sequentielle(notes, 17))
print(recherche_dichotomique(notes_triees, 17))
```

## TP 3 — Récursivité

```python
def factorielle(n):
    """Calcule n! recursivement."""
    if n <= 1:   # cas de base — obligatoire !
        return 1
    return n * factorielle(n - 1)

print(factorielle(5))
print(factorielle(0))
```

```python
def fibonacci(n):
    if n <= 1: return n
    return fibonacci(n-1) + fibonacci(n-2)

for i in range(10):
    print(fibonacci(i), end=' ')
```

```python
# Comparaison recursif vs iteratif
def fibonacci_iter(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

import time
t0 = time.time(); fibonacci(35);      print(f'Recursif  : {time.time()-t0:.3f}s')
t0 = time.time(); fibonacci_iter(35); print(f'Iteratif  : {time.time()-t0:.6f}s')
```

## TP 4 — Fichiers CSV

```python
import csv, os

# Creer le fichier eleves.csv pour les tests
contenu = """nom,prenom,classe,note_maths,note_info
Dupont,Alice,1NSI1,15,18
Martin,Bob,1NSI2,12,14
Bernard,Clara,1NSI1,17,16
Thomas,David,1NSI2,9,11
Leroy,Emma,1NSI1,14,19
"""
with open('eleves.csv', 'w', encoding='utf-8') as f:
    f.write(contenu)
print("Fichier eleves.csv cree")
```

```python
with open('eleves.csv', newline='', encoding='utf-8') as f:
    lecteur = csv.DictReader(f)
    for ligne in lecteur:
        print(ligne['prenom'], ligne['nom'], ligne['note_info'])
```

```python
def lire_csv(chemin):
    """Retourne une liste de dicts depuis un CSV."""
    with open(chemin, newline='', encoding='utf-8') as f:
        return list(csv.DictReader(f))

eleves = lire_csv('eleves.csv')
print(len(eleves), 'eleves', eleves[0])
```

```python
# Traitement
notes_info = [int(e['note_info']) for e in eleves]
print(f'Moy info : {sum(notes_info)/len(notes_info):.2f}')
admis = [e for e in eleves if int(e['note_info']) >= 10]
print(f'{len(admis)} admis en info')
```

```python
def ecrire_csv(chemin, eleves, champs):
    with open(chemin, 'w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=champs, extrasaction='ignore')
        writer.writeheader()
        writer.writerows(eleves)

champs = ['nom', 'prenom', 'note_info']
ecrire_csv('admis_info.csv', admis, champs)
print("Fichier admis_info.csv cree")

# Verification
resultat = lire_csv('admis_info.csv')
print(len(resultat), 'eleves dans admis_info.csv')
```
