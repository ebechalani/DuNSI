## TP 1 - Fonctions

```python
def saluer(prenom):
    """Affiche un message de bienvenue."""
    print(f'Bonjour {prenom} !')

saluer('Alice')
saluer('Bob')
```

```python
def carre(n):
    """Retourne le carre de n."""
    return n * n

print(carre(5))
print(carre(carre(3)))   # predisez avant d'executer
```

```python
def mention(note, seuil=10):
    return 'Admis' if note >= seuil else 'Ajourne'

print(mention(14))
print(mention(14, 15))
print(mention(note=14, seuil=12))
```

```python
def statistiques(notes):
    """Retourne (min, max, moyenne) d'une liste."""
    return min(notes), max(notes), sum(notes)/len(notes)

notes = [12, 15, 8, 17, 11]
mini, maxi, moy = statistiques(notes)
print(f'Min={mini}  Max={maxi}  Moy={moy:.2f}')
```

## TP 4 - Compréhensions

```python
notes = [12, 15, 8, 17, 11, 14, 6]
carres = [n**2 for n in notes]
admis  = [n for n in notes if n >= 10]
pct    = [round(n/20*100, 1) for n in notes]
print(admis)
print(pct)
```

```python
noms = ['Alice', 'Bob', 'Clara']
vals = [15, 12, 17]
dico = {nom: note for nom, note in zip(noms, vals)}
print(dico)
```
