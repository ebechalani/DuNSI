## Étape 1 — Le tableau 1D (un vecteur)
Un tableau numpy 1D ressemble à une liste, mais les opérations s'appliquent à TOUS les éléments d'un coup (sans boucle).

```python
import numpy as np
notes = np.array([12, 15, 8, 17, 11])
print(notes[0], notes[-1])   # premier, dernier
print(notes[1:3])            # tranche [15 8]
print(notes + 1)             # +1 à TOUS les éléments
print(notes.mean(), notes.max())

# À toi : crée le tableau des carrés de 1 à 5 -> [1 4 9 16 25]
# carres = np.array([...]) ** 2
```

## Étape 2 — Le tableau 2D (une matrice)
Un tableau 2D, c'est des lignes ET des colonnes. On accède à une case avec [ligne, colonne].

```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m.shape)     # (2, 3) : 2 lignes, 3 colonnes
print(m[0, 1])     # ligne 0, colonne 1  -> 2
print(m[1])        # toute la ligne 1    -> [4 5 6]
print(m[:, 0])     # toute la colonne 0  -> [1 4]

# À toi : affiche la somme de chaque ligne (indice : m.sum(axis=1))
```

## Étape 3 — Créer et transformer des tableaux 2D
numpy sait fabriquer des tableaux et les transformer globalement.

```python
zeros = np.zeros((3, 4))                 # tableau 2D rempli de 0
print(zeros)

degrade = np.tile(np.linspace(0, 255, 5, dtype=int), (3, 1))
print(degrade)        # chaque ligne va de 0 à 255
print(255 - degrade)  # on inverse toutes les valeurs (négatif)
```

## Étape 4 — Une image EST un tableau 2D !
Une image en niveaux de gris = un tableau 2D de pixels (0 = noir, 255 = blanc). On l'affiche avec matplotlib.

```python
import matplotlib.pyplot as plt
image = np.tile(np.linspace(0, 255, 100, dtype=int), (100, 1))
plt.imshow(image, cmap='gray')
plt.title('Un dégradé = une image en niveaux de gris')
plt.show()
```

## Étape 5 — Manipuler l'image
Maintenant qu'une image est un tableau, on la transforme comme un tableau : négatif, seuillage.

```python
negatif = 255 - image
seuil = np.where(image > 128, 255, 0)   # binarise : >128 -> blanc, sinon noir

fig, ax = plt.subplots(1, 3, figsize=(11, 3))
ax[0].imshow(image, cmap='gray');   ax[0].set_title('original')
ax[1].imshow(negatif, cmap='gray'); ax[1].set_title('négatif')
ax[2].imshow(seuil, cmap='gray');   ax[2].set_title('seuillage')
plt.show()

# À toi : crée une image 100x100 dont la moitié gauche est noire et la droite blanche
```
