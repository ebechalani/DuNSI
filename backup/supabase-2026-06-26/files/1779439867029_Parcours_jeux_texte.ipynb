# Parcours guidé — Programmer un jeu texte

Les briques d'un jeu : mot secret, affichage masqué, boucle de jeu.

## 🎮 Jeu — choisir un mot au hasard

```python
import random
mot = random.choice(['python', 'classe', 'boucle'])
print('Le mot a', len(mot), 'lettres')
```

## Étape 1 — Mémoriser les lettres trouvées (ensemble)

```python
trouvees = set()
trouvees.add('p')
trouvees.add('y')
print('p' in trouvees)   # True
print('z' in trouvees)   # False
```

## Étape 2 — Afficher le mot masqué

```python
mot = 'python'
trouvees = {'p', 'y', 't'}
affichage = ''.join(c if c in trouvees else '_' for c in mot)
print(affichage)   # pyt__n -> pyt___
```

## Étape 3 — La boucle de jeu

```python
mot = 'python'
trouvees = set()
essais = 6
while essais > 0:
    aff = ''.join(c if c in trouvees else '_' for c in mot)
    print(aff, '  essais :', essais)
    if '_' not in aff:
        print('Gagné !'); break
    lettre = input('Lettre : ')
    if lettre in mot:
        trouvees.add(lettre)
    else:
        essais -= 1
```
