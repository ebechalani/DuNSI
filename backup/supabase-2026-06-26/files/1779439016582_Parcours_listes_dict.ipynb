## Étape 2 — Les dictionnaires
Clé -> valeur ; accès par NOM de clé.

```python
eleve = {'nom': 'Dupont', 'note': 18}
print(eleve['nom'])
print(eleve.get('age', 'inconnu'))   # défaut si absent
eleve['classe'] = '1NSI1'            # ajout
for cle, val in eleve.items():
    print(cle, ':', val)
```

## Étape 3 — Deux listes -> un dictionnaire
dict(zip(...)) apparie deux listes.

```python
pays = ['Panama', 'Liberia', 'Allemagne']
navires = [663, 477, 273]
d = dict(zip(pays, navires))
print(d)
print(d['Panama'])   # 663
# À toi : refais-le avec une boucle (sans zip)
```

## Étape 4 — Croiser deux dictionnaires par clé
Le point clé : les deux dicts ne sont PAS dans le même ordre, mais on retrouve chaque valeur par sa clé.

```python
conteneurs = {'Panama': 663, 'Allemagne': 273}
tonnage    = {'Allemagne': 11646, 'Panama': 23282}   # ordre différent !
for pays in conteneurs:
    moyenne = tonnage[pays] / conteneurs[pays]
    print(pays, round(moyenne, 1))
```
