## 🎮 Jeu — dessiner un carré

```python
import turtle
t = turtle.Turtle()
for _ in range(4):
    t.forward(80)
    t.left(90)
turtle.done()
```

## Étape 1 — Un polygone régulier
Un polygone à n côtés : avancer puis tourner de 360/n, n fois.

```python
import turtle
t = turtle.Turtle()

def polygone(t, n, cote):
    for _ in range(n):
        t.forward(cote)
        t.left(360 / n)

polygone(t, 6, 70)   # hexagone
turtle.done()
```

## Étape 2 — Une rosace (répéter en tournant)

```python
import turtle
t = turtle.Turtle(); t.speed(0)
for _ in range(36):
    polygone(t, 4, 80)   # un carré
    t.left(10)           # tourner un peu
turtle.done()
```

## Étape 3 — La récursivité géométrique : un arbre fractal

```python
import turtle
t = turtle.Turtle(); t.speed(0); t.left(90)

def arbre(t, longueur):
    if longueur < 10:
        return
    t.forward(longueur)
    t.left(30);  arbre(t, longueur * 0.7)
    t.right(60); arbre(t, longueur * 0.7)
    t.left(30);  t.backward(longueur)

arbre(t, 80)
turtle.done()
```
