# Parcours guidé — Le backtracking (retour sur trace)

Explorer, et revenir en arrière quand on est bloqué : la sortie d'un labyrinthe.

## 🎮 Jeu — afficher un labyrinthe (0 = libre, 1 = mur)

```python
laby = [[0, 1, 0],
        [0, 0, 0],
        [1, 1, 0]]
for ligne in laby:
    print(''.join('#' if c else '.' for c in ligne))
```

## Étape 1 — Une case est-elle praticable ?

```python
def valide(laby, x, y):
    return (0 <= x < len(laby) and 0 <= y < len(laby[0]) and laby[x][y] == 0)

print(valide(laby, 0, 0))   # True (libre)
print(valide(laby, 0, 1))   # False (mur)
```

## Étape 2 — Explorer et REVENIR EN ARRIÈRE
On tente une direction ; si c'est une impasse, on retire la case (backtrack) et on essaie une autre.

```python
def resoudre(laby, x, y, chemin):
    if not valide(laby, x, y) or (x, y) in chemin:
        return False
    chemin.append((x, y))
    if (x, y) == (len(laby) - 1, len(laby[0]) - 1):
        return True                      # sortie atteinte
    for dx, dy in [(1, 0), (0, 1), (-1, 0), (0, -1)]:
        if resoudre(laby, x + dx, y + dy, chemin):
            return True
    chemin.pop()                         # impasse -> backtrack
    return False

chemin = []
resoudre(laby, 0, 0, chemin)
print(chemin)
```
