# Structures linéaires
    
Les **listes**, **piles** et **files** sont des structures de données à une dimension permettant de stocker **de manière linéaire** une suite finie de données. On les dénomme parfois **structures linéaires**.
   
### Les listes python   
    
Une listes python est une collection de variables, ordonnées.
    
```python
ma_liste = [1, 23, 45, 14, 63 ,478, 2, 3, 456, 0]   
```    
    
**Interface :** actions possibles sur la structure de donnée.
    
- **`ma_liste.append(x)`** : ajoute **`x`** a la fin ;  
- **`ma_liste.pop()`** : retire et renvoi le dernier élément ;   
- **`ma_liste.insert(n, x)`** : insert l’élément **`x`** à l’emplacement **`n`** ;  
- **`ma_liste.remove(x)`** : supprime la première occurrence de l’élément **`x`** ; 
- **`ma_liste.reverse()`** : inverse l'ordre des valeurs stockées dans la liste ;  
- **`ma_liste.sort()`** : trie la liste de valeur dans l'ordre croissant ; 
- **`ma_liste.index(x)`** : renvoie l'indice la première occurrence de l’élément **`x`** dans la liste ;    
- …
   
    
### Types abstraits et représentation concrète des données   
    
- Les **listes**, les **piles** ou les **files** sont des "**vues de l'esprit**" présent uniquement dans la tête des informaticiens, on dit que ce sont des **types abstraits**.  
 
- Vu de l’utilisateur, **un type abstrait est accessible par son interface**.  

- L’**implémentation** d’un type de donnée est sa **traduction** dans un **langage** compréhensible par l’ordinateur (Python, Java, C, …). Elle dépend du langage utilisé.  

- Certains types abstraits ne sont pas forcément implémentés dans un langage donné, si le programmeur veut utiliser ce type abstrait, il faudra qu'il le programme par lui-même en utilisant les "outils" fournis par son langage de programmation. Par exemple, pas de pile, ni de file en python.  
    
    
## Les piles   

L’image la plus simple pour se représenter une pile est une **pile d’assiette** comme vous en avez tous une dans votre cuisine : 
    
   

Les piles sont basées sur le principe **LIFO** (**Last In First Out** : le dernier rentré sera le premier à sortir). On retrouve souvent ce principe LIFO en informatique.
    
### Interfaces des piles
    
On appelle les opérations de bases applicables a une structure de donnée des **primitives**.

Sur une pile, l’interface permet uniquement :

- d'**empiler** un nouvel élément (ajouter un élément au sommet de la pile) ;

- de **dépiler** le dernier élément saisi ;

- de **consulter la valeur de l'élément au sommet de la pile** sans le dépiler ;

- de **tester si la pile est vide** ;

- de **connaître le nombre d'éléments dans la pile**.

Vous voyez donc qu'il n'est pas possible d'accéder directement au n-ième élément d'une pile : Il faudrait dépiler les n-1 premiers éléments et de ce fait, détruire partiellement notre pile.

### Utilisation des piles      
    
Les piles sont extrêmement utiles en informatique et vous les utilisez quotidiennement, parfois même sans vous en rendre compte :

- **La fonction annuler (Ctrl-Z) de votre traitement de textes** par exemple est une pile : Quand vous tapez Ctrl-Z, vous annulez la dernière opération effectuée. Quand vous faites une nouvelle opération, celle-ci est mémorisée au sommet de la pile. Vous ne pouvez pas annuler l'avant dernière opération sauf à annuler la dernière.   

- **Le bouton retour de votre navigateur internet** fonctionne également à l'aide d'une pile. Les pages web consultées lors de votre navigation sur une page sont empilées et le bouton retour permet d'accéder à la dernière page présente sur la pile.

### Utilisation des piles : la notation polonaise inversée
    
**La notation polonaise inverse (NPI)** (en anglais RPN pour Reverse Polish Notation), également connue sous le nom de notation post-fixée, permet d'écrire de façon non ambiguë les formules arithmétiques sans utiliser de parenthèses.   

Dérivée de la notation polonaise présentée en 1924 par le mathématicien polonais Jan Łukasiewicz, elle s’en différencie par l’ordre des termes, les opérandes y étant présentés avant les opérateurs et non l’inverse.     
    
L’expression **2 + 3** devient en NPI : **2 3 +**.   
    
    
***Autres exemples :***  
    - **2 + 6 − 1** s’écrit **2 6 + 1 -**   
    - **5 ∗ 3 + 4** s’écrit **5 3 * 4 +**   
    - **((1 + 2) ∗ 4) + 3** s’écrit **1 2 + 4 * 3 +**      
    
Évaluer une expression post-fixée est facile, il suffit de lire l’expression de gauche à droite et d’appliquer chaque opérateur aux deux opérandes qui le précèdent. Si l’opérateur n’est pas le dernier symbole on replace le résultat intermédiaire dans l’expression et on recommence avec l’opérateur suivant.   
    
Le but de l’exercice est de réaliser en Python une calculatrice simple, capable d’évaluer une formule en NPI et de retourner le résultat arithmétique.   
    
La réalisation d’une telle calculatrice se fera à l’aide d’une pile.      
    

***Algorithme de calcul d'une expression NPI :***
    
- Lire un par un les caractères de l’expression.
- Si le caractère lu est un opérande(nombre) alors on l’empile.
- Si le caractère lu est un opérateur, alors on dépile les deux éléments se trouvant en haut de la pile, on calcule le résultat en appliquant l’opérateur sur les deux opérandes dépilés et on empile le résultat.
- Une fois tous les caractères lus, la pile ne contient qu’un seul élément qui correspond au résultat final.

## Les files    
    
Dans une file, les éléments sont placés les uns à cotés des autres comme dans une pile, à la différence que l'on sort les éléments du plus ancien vers le plus récent. Cela correspond à ce qui se passe dans une **file d'attente** :      
    
    
    
Les files sont basées sur le principe **FIFO** (**First In First Out** : le premier qui est rentré sera le premier à sortir. Ici aussi, on retrouve souvent ce principe FIFO en informatique. Par exemple dans les files d'impression où le premier document envoyé à l'imprimante sera le premier document à être imprimé.  
  
### Interfaces des files   
    
Les opérations permises sur les files sont similaires aux piles. Il est possible :   

- d'**emfiler** un nouvel élément (ajouter un élément à la fin de la file)

- de **défiler** le premier élément saisi (celui au début de la file)

- de **consulter la valeur du premier élément** au début de la file sans le sortir de la file

- de **tester si la file est vide**

- de **connaître le nombre d'éléments** dans la file.

Vous voyez donc qu'il n'est **pas possible** d'accéder directement au **n-ième** élément d'une file : Il faudrait défiler les n-1 premiers éléments et de ce fait, détruire partiellement notre file.

### Utilisation des files : Ordonnanceur de processus : tourniquet (round-robin).  

Un système d'exploitation doit partager le processeur entre plusieurs processus. Un processus peut être dans différents états :
    
    
    
Le **tourniquet (round-robin)** est la politique la plus simple :   
    
- les processus prêts sont rangés dans une **file FIFO** ;   
- l'ordonnanceur défile le processus en tête et lui accorde le processeur pendant un quantum de temps fixe (par ex. 4 cycles CPU) ;   
- s'il termine avant, parfait ; sinon il est préempté : il est remis en fin de file (comme s'il redemandait son tour) ;   
- on défile le suivant, et ainsi de tour en tour d'où le nom de "tourniquet".   

    
    
C'est une application directe de la propriété FIFO : personne ne monopolise le CPU, chaque processus est servi de façon équitable.
