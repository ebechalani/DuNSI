# Algorithmes Diviser pour régner  

### Principe    
    
Le **diviser pour régner** est une **méthode algorithmique** basée sur le principe suivant :  

Prenons **un problème complexe à résoudre** par taille. l'idée est de le **diviser** en une multitude de **petits problèmes**, dans l'objectif que ces "petits problèmes" seront plus simples à résoudre que le problème original. Une fois les petits problèmes résolus, on **recombine** les "petits problèmes résolus" afin d'obtenir la solution du problème de départ.  
    
    
Ce sont généralement des **algorithmes récursifs**.
  
### Algorithme
    
Le paradigme "diviser pour régner" repose donc sur 3 étapes :   

- **DIVISER :** le problème d'origine est divisé en un certain nombre de sous-problèmes

- **RÉGNER :** on résout les sous-problèmes (les sous-problèmes sont plus faciles à résoudre que le problème d'origine)

- **COMBINER :** les solutions des sous-problèmes sont combinées afin d'obtenir la solution du problème d'origine.

# Le tri fusion
    
### Un autre exemple d’algorithme diviser pour régner.  

Le tri – fusion a été inventé par John von Neumann en 1945.  
 
L'opération principale de l'algorithme est la fusion, qui consiste à réunir deux listes triées en une seule. L'efficacité de l'algorithme vient du fait que deux listes triées peuvent être fusionnées en temps linéaire.

  
    
    
### Principe

Cet algorithme repose sur deux fonctions :  
  
    
- **`fusion`** : recombine deux listes triée en une seule (également triée évidemment).  Il s’agit des phases **REGNER** et **COMBINER**.

Exemple :  
```python    
>>> fusion([1, 3], [2, 4]) 
[1, 2, 3, 4]
```    

- **`tri_Fusion`** : C’est une fonction récursive. Cette fonction coupe en deux la liste a trier. Elle applique la fonction triFusion (elle-même) sur chaque moitié de liste et recombine le résultat. Il s’agit de la phase **DIVISER**.
