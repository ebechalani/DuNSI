## 🎮 Jeu d'introduction — que calcule cette fonction ?

```python
def mystere(n):
    if n == 0:
        return 1
    return n * mystere(n - 1)

print(mystere(4))   # ?  Devine AVANT d'exécuter
```

## Étape 1 — Cas de base + appel récursif
Une fonction récursive s'appelle elle-même sur un problème plus petit, jusqu'à un cas de base qui arrête tout.

```python
def compte(n):
    if n == 0:            # cas de base
        print('Décollage !')
        return
    print(n)
    compte(n - 1)        # appel sur plus petit

compte(3)
```

## Étape 2 — La factorielle
n! = n × (n-1)!, avec 0! = 1.

```python
def factorielle(n):
    if n <= 1:
        return 1
    return n * factorielle(n - 1)

print(factorielle(5))   # 120
# À toi : déroule à la main factorielle(4)
```

## Étape 3 — La suite de Fibonacci
F(n) = F(n-1) + F(n-2), avec F(0)=0 et F(1)=1.

```python
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

for i in range(10):
    print(fib(i), end=' ')
```

## Étape 4 — Récursif vs itératif (lent vs rapide)

```python
import time

def fib_iter(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

t0 = time.time(); fib(30);      print('récursif :', round(time.time() - t0, 3), 's')
t0 = time.time(); fib_iter(30); print('itératif :', time.time() - t0, 's')
```
