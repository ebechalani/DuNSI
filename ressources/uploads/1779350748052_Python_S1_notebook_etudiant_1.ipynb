```python
# Accumulation : somme 1 + 2 + ... + 100
total = 0
for i in range(1, 101):
    total += i
print(f'Somme de 1 à 100 : {total}')

# Vérification avec la formule mathématique
n = 100
formule = n * (n + 1) // 2
print(f'Formule n(n+1)/2  : {formule}')
print(f'Égaux ? {total == formule}')
```

### 4.2 — Boucle `while` : jeu du deviné

> `while True: ... break` : motif classique pour une boucle jusqu'à condition.

```python
import random

secret    = random.randint(1, 20)
tentative = 0

while True:
    guess = int(input('Devinez un nombre entre 1 et 20 : '))
    tentative += 1
    if guess == secret:
        print(f' Bravo ! Trouvé en {tentative} coup(s) !')
        break
    elif guess < secret:
        print(' Trop petit')
    else:
        print(' Trop grand')
```
