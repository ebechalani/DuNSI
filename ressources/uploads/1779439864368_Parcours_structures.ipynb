Les structures de données au programme : pile (LIFO), file (FIFO), arbre.

## 🎮 Jeu — la pile d'assiettes

## Étape 1 — La pile (LIFO : dernier entré, premier sorti)

```python
pile = []
pile.append(1); pile.append(2); pile.append(3)
print(pile.pop())   # 3
print(pile)         # [1, 2]
```

## Étape 2 — La file (FIFO : premier entré, premier sorti)

```python
from collections import deque
file = deque()
file.append(1); file.append(2); file.append(3)
print(file.popleft())   # 1
```

## Étape 3 — Application : parenthésage équilibré (avec une pile)

```python
def equilibre(expr):
    pile = []
    paires = {')': '(', ']': '[', '}': '{'}
    for c in expr:
        if c in '([{':
            pile.append(c)
        elif c in ')]}':
            if not pile or pile.pop() != paires[c]:
                return False
    return len(pile) == 0

print(equilibre('(a[b]{c})'))   # True
print(equilibre('(a[b)'))       # False
```

## Étape 4 — L'arbre (un nœud, deux enfants)

```python
class Noeud:
    def __init__(self, v):
        self.v = v
        self.g = None
        self.d = None

racine = Noeud(5)
racine.g = Noeud(3)
racine.d = Noeud(8)
print(racine.v, '->', racine.g.v, 'et', racine.d.v)
```
