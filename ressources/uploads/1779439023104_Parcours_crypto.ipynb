## Étape 1 — Parcourir une chaîne
Une chaîne est une suite de caractères. ord donne le code, chr l'inverse.

```python
mot = 'BONJOUR'
for c in mot:
    print(c, ord(c))
```

## Étape 2 — Décaler une lettre (modulo 26)
On ramène la lettre entre 0 et 25, on décale, le modulo 26 gère le repli (Z -> A).

```python
c = 'A'
decalage = 3
nouveau = chr((ord(c) - ord('A') + decalage) % 26 + ord('A'))
print(nouveau)   # D
# À toi : décale 'Z' de 1 -> doit donner 'A'
```

## Étape 3 — Chiffrer un mot entier

```python
def chiffrer(mot, d):
    res = ''
    for c in mot:
        res += chr((ord(c) - ord('A') + d) % 26 + ord('A'))
    return res

print(chiffrer('BONJOUR', 3))   # ERQMRXU
```
