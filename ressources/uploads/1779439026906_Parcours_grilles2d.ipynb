## 🎮 Jeu d'introduction — affiche une grille

```python
grille = [[0, 1, 0],
          [1, 1, 1],
          [0, 1, 0]]
for ligne in grille:
    print(ligne)
```

## Étape 1 — Une grille = une liste de listes
Chaque sous-liste est une ligne ; grille[i][j] = case ligne i, colonne j.

```python
grille = [[0] * 3 for _ in range(3)]   # grille 3x3 de zéros
grille[1][1] = 1                       # case du centre
print(grille)
print(grille[1][1])                    # 1
```

## Étape 2 — Parcourir et afficher joliment

```python
for i in range(len(grille)):
    for j in range(len(grille[0])):
        print('#' if grille[i][j] else '.', end='')
    print()
# À toi : mets toute la première ligne à 1 et réaffiche
```

## Étape 3 — Compter les voisins d'une case
Les 8 voisines autour de (i, j), en restant dans la grille.

```python
def voisins(g, i, j):
    n = 0
    for di in (-1, 0, 1):
        for dj in (-1, 0, 1):
            if (di, dj) != (0, 0):
                x, y = i + di, j + dj
                if 0 <= x < len(g) and 0 <= y < len(g[0]):
                    n += g[x][y]
    return n

print(voisins(grille, 1, 1))
```
