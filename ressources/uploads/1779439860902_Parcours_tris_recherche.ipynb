# Parcours guidé — Algorithmes de tri et recherche

De l'échange de deux cases aux tris, puis à la recherche dichotomique.

## 🎮 Jeu — l'échange (la base de tout tri)

```python
L = [5, 2, 8]
L[0], L[1] = L[1], L[0]   # échange
print(L)   # prédis avant d'exécuter
```

## Étape 1 — Trouver le minimum

```python
L = [5, 2, 8, 1, 9]
imin = 0
for j in range(1, len(L)):
    if L[j] < L[imin]:
        imin = j
print('minimum :', L[imin], 'en position', imin)
```

## Étape 2 — Tri par sélection
À chaque tour : placer le minimum du reste à sa place.

```python
def tri_selection(L):
    L = L.copy()
    for i in range(len(L) - 1):
        imin = i
        for j in range(i + 1, len(L)):
            if L[j] < L[imin]:
                imin = j
        L[i], L[imin] = L[imin], L[i]
    return L

print(tri_selection([5, 2, 8, 1, 9, 3]))
```

## Étape 3 — Tri par insertion
Insérer chaque élément à sa place dans la partie déjà triée.

```python
def tri_insertion(L):
    L = L.copy()
    for i in range(1, len(L)):
        cle = L[i]; j = i - 1
        while j >= 0 and L[j] > cle:
            L[j + 1] = L[j]; j -= 1
        L[j + 1] = cle
    return L

print(tri_insertion([5, 2, 8, 1, 9, 3]))
```

## Étape 4 — Recherche dichotomique (liste TRIÉE)

```python
def dichotomie(L, v):
    g, d = 0, len(L) - 1
    while g <= d:
        m = (g + d) // 2
        if L[m] == v: return m
        elif L[m] < v: g = m + 1
        else: d = m - 1
    return -1

print(dichotomie([1, 2, 3, 5, 8, 9], 8))
```
