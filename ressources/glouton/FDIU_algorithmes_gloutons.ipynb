# Les algorithmes gloutons #  
    
## problèmes d’optimisation   
    
Les algorithmes gloutons sont utilisés pour résoudre des problèmes d’optimisation : **Trouver la meilleure solution à un problème qui en offre plusieurs**.  

***Exemples :***   
    
Rendre la monnaie avec 
une minimum de pièce.
    
    

Problème du sac à dos : Votre sac à dos ne peut 
contenir que 5 kg, choisir comment le remplir 
avec la plus grande valeur d’objet sans dépasser ce poids.


 
    
Des conférenciers sont invités à présenter leurs 
exposés dans une salle. Mais leurs disponibilités 
ne leur permettent d’intervenir qu’à des horaires 
bien définis. Le problème est de construire un 
planning d’occupation de la salle avec le plus 
grand nombre de conférenciers.

    
    
## Solutions algorithmiques    
    
Il existe deux grandes familles d'algorithme qui permettenet de résoudre ce type de problème :  
    
- **Les algorithmes gloutons :** Cette solution consiste a appliquer une règle, toujours la même, en espérant qu'elle nous conduise a une solution optimale. Ces algorithmes n'évaluent pas la solution trouvée, et ne reviennent jamais en arrière. Cela conduit a des fonctions généralement simples. Rien ne garantit, par contre, que le résultat trouvé soit optimal. 
    
- **La recherche exhaustive :** Consiste a chercher toutes les solutions possibles a un problème, puis, à trier ces solutions pour ne retenir que la meilleure. Cette solution nécessite énormément de calcul mais le résultat trouvé est optimal.

??? info Les secrets de sorted
**sorted** renvoie une copie triée de la liste passée en argument, celle-ci n'est pas modifiée.

```python
>>> lst1 = [125, 457, 148, 698, 12, 1, 789, 14]
>>> lst2 = sorted(lst1)
>>> lst1
[125, 457, 148, 698, 12, 1, 789, 14]
>>> lst2
[1, 12, 14, 125, 148, 457, 698, 789]
```

l'argument otionnel **reverse** permet de triée dans le sens décroissant.

```python
>>> lst1 = [125, 457, 148, 698, 12, 1, 789, 14]
>>> lst2 = sorted(lst1)
>>> lst1
[125, 457, 148, 698, 12, 1, 789, 14]
>>> lst2
[1, 12, 14, 125, 148, 457, 698, 789]
```

```python
>>> lst1 = [125, 457, 148, 698, 12, 1, 789, 14]
>>> lst3 = sorted(lst1, reverse=True)
>>> lst3
[789, 698, 457, 148, 125, 14, 12, 1]
```
Si la liste contient des itérables, le tri se fait sur le premier élément de chacun de ces itérables.   
Pour trier sur un autre élément, il faut passer une fonction a sorted qui renvoi l'élément qui doit servir de clé au tri. Cela est généralement fait avec une lambda fonction.  

Exemple, une liste de tuple.   


```python
>>> list4 = [(999, 90, 585), (20, 462, 60), (139, 71, 343)]
>>> sorted(lst1) # tri sur le premier élément
[(20, 462, 60), (139, 71, 343), (999, 90, 585)]
>>> sorted(list4, key=lambda x: x[1]) # tri sur le second élément
[(139, 71, 343), (999, 90, 585), (20, 462, 60)]
```

## Recherche des solutions exhaustives
    
C'est l'algorithme concurent des algorithmes gloutons pour les problèmes d'optimisation. Il consiste a rechercher toutes les solutions possibles et a sélectionner la meilleure. Ce sont généralement des algorithmes récurcifs.  
    
### Algorithme   
Le tableau des conférenciers est classé dans l'ordre croissant des heures de début.    
planning2(tab_inter, debut=0, i=0):  

Meilleur planning commençant à début avec les conférences i et plus    
- si il n'y a plus de conférence disponibles :
    - renvoyer un planning vide
- si la conférence i commence avant début :
    - rechercher une solution avec i + 1
- sinon :
    - solution 1 : choisir la conférence i et poursuivre le planning avec la conférence i + 1, qui débute après i
    - solution 2 : ne pas choisir la conférence i et rechercher une solution avec i + 1
    - renvoyer la meilleure des deux solutions.
